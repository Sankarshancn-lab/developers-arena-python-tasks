```python
# ==========================================
# WEEK 5 - CUSTOMER SALES ANALYSIS
# DAY 1: DATA LOADING & EXPLORATION
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

# Display configuration
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

print("=" * 60)
print("       CUSTOMER SALES ANALYSIS PROJECT")
print("=" * 60)

print("\nAll libraries imported successfully!")
```

```python
# Load datasets

sales_df = pd.read_csv("sales_data.csv")

customer_df = pd.read_csv("customer_churn.csv")

print("Datasets loaded successfully!")

print("\nSales Dataset Shape:", sales_df.shape)
print("Customer Dataset Shape:", customer_df.shape)
```

```python
# ==========================================
# STEP 3: DATASET STRUCTURE EXPLORATION
# ==========================================

print("=" * 60)
print("SALES DATASET INFORMATION")
print("=" * 60)

print("\nFirst 5 Records:")
display(sales_df.head())

print("\nDataset Information:")
sales_df.info()

print("\nColumn Names:")
print(sales_df.columns.tolist())

print("\nDataset Dimensions:")
print("Rows:", sales_df.shape[0])
print("Columns:", sales_df.shape[1])


print("\n" + "=" * 60)
print("CUSTOMER DATASET INFORMATION")
print("=" * 60)

print("\nFirst 5 Records:")
display(customer_df.head())

print("\nDataset Information:")
customer_df.info()

print("\nColumn Names:")
print(customer_df.columns.tolist())

print("\nDataset Dimensions:")
print("Rows:", customer_df.shape[0])
print("Columns:", customer_df.shape[1])
```

```python
# ==========================================
# STEP 4: STATISTICAL SUMMARY
# ==========================================

print("SALES DATASET STATISTICAL SUMMARY")

display(sales_df.describe())

print("\nCUSTOMER DATASET STATISTICAL SUMMARY")

display(customer_df.describe())
```

```python
# ==========================================
# STEP 5: DATA QUALITY CHECK
# ==========================================

print("SALES DATASET")

print("\nMissing Values:")
print(sales_df.isnull().sum())

print("\nDuplicate Records:")
print(sales_df.duplicated().sum())


print("\n" + "=" * 50)

print("CUSTOMER DATASET")

print("\nMissing Values:")
print(customer_df.isnull().sum())

print("\nDuplicate Records:")
print(customer_df.duplicated().sum())
```

```python
# ==========================================
# STEP 6: CATEGORICAL DATA EXPLORATION
# ==========================================

print("=" * 60)
print("CATEGORICAL DATA ANALYSIS")
print("=" * 60)

# Product distribution
print("\nProduct Distribution:")
print(sales_df['Product'].value_counts())

# Regional distribution
print("\nRegional Distribution:")
print(sales_df['Region'].value_counts())

# Unique customers
print("\nTotal Unique Customers:")
print(sales_df['Customer_ID'].nunique())

# Customer transaction frequency
print("\nTop 10 Customers by Transaction Count:")
print(sales_df['Customer_ID'].value_counts().head(10))

# Churn distribution
print("\nCustomer Churn Distribution:")
print(customer_df['Churn'].value_counts())
```

```python
# ==========================================
# STEP 7: INITIAL BUSINESS PERFORMANCE
# ==========================================

total_revenue = sales_df['Total_Sales'].sum()

total_transactions = len(sales_df)

total_customers = sales_df['Customer_ID'].nunique()

average_order_value = sales_df['Total_Sales'].mean()

total_quantity = sales_df['Quantity'].sum()

average_product_price = sales_df['Price'].mean()

print("=" * 60)
print("       INITIAL SALES PERFORMANCE REPORT")
print("=" * 60)

print(f"\nTotal Revenue: {total_revenue:,.2f}")
print(f"Total Transactions: {total_transactions:,}")
print(f"Total Unique Customers: {total_customers:,}")
print(f"Average Order Value: {average_order_value:,.2f}")
print(f"Total Units Sold: {total_quantity:,}")
print(f"Average Product Price: {average_product_price:,.2f}")

print("\n" + "=" * 60)
print("INITIAL BUSINESS ANALYSIS COMPLETED")
print("=" * 60)
```

```python
# ==========================================
# DAY 2: DATA CLEANING & PREPARATION
# STEP 1: DATA TYPE INSPECTION
# ==========================================

print("=" * 60)
print("SALES DATA TYPE INSPECTION")
print("=" * 60)

print(sales_df.dtypes)

print("\nCUSTOMER DATA TYPE INSPECTION")
print("=" * 60)

print(customer_df.dtypes)

print("\nDataset Memory Usage:")

print("Sales Dataset:",
      sales_df.memory_usage(deep=True).sum() / 1024, "KB")

print("Customer Dataset:",
      customer_df.memory_usage(deep=True).sum() / 1024, "KB")
```

```python
# Check duplicate records

print("SALES DATASET")

print("Total duplicates:", sales_df.duplicated().sum())

print("\nCUSTOMER DATASET")

print("Total duplicates:", customer_df.duplicated().sum())

print("\nDuplicate Customer IDs:",
      customer_df['CustomerID'].duplicated().sum())
```

```python
# Validate sales calculations

sales_df['Calculated_Sales'] = (
    sales_df['Quantity'] * sales_df['Price']
)

sales_df['Sales_Mismatch'] = (
    sales_df['Total_Sales'] != sales_df['Calculated_Sales']
)

print("Total Sales Records:", len(sales_df))

print("Incorrect Sales Calculations:",
      sales_df['Sales_Mismatch'].sum())

print("\nSales Validation:")
display(
    sales_df[
        ['Quantity', 'Price', 'Total_Sales',
         'Calculated_Sales', 'Sales_Mismatch']
    ].head(10)
)
```

```python
# ==========================================
# STEP 4: DATETIME CONVERSION
# ==========================================

sales_df['Date'] = pd.to_datetime(
    sales_df['Date'],
    errors='coerce'
)

print("Date conversion completed!")

print("\nDate Column Data Type:")
print(sales_df['Date'].dtype)

print("\nDate Range:")
print("Start Date:", sales_df['Date'].min())
print("End Date:", sales_df['Date'].max())

print("\nInvalid Dates:", sales_df['Date'].isnull().sum())
```

```python
# Extract date components

sales_df['Year'] = sales_df['Date'].dt.year

sales_df['Month'] = sales_df['Date'].dt.month

sales_df['Day'] = sales_df['Date'].dt.day

sales_df['Month_Name'] = sales_df['Date'].dt.month_name()

sales_df['Day_Name'] = sales_df['Date'].dt.day_name()

print("Date components extracted successfully!")

display(
    sales_df[
        ['Date', 'Year', 'Month', 'Day',
         'Month_Name', 'Day_Name']
    ].head(10)
)
```

```python
# ==========================================
# TEXT DATA CLEANING
# ==========================================

# Remove leading and trailing spaces
sales_df['Product'] = (
    sales_df['Product'].str.strip().str.title()
)

sales_df['Region'] = (
    sales_df['Region'].str.strip().str.title()
)

# Standardize customer IDs
sales_df['Customer_ID'] = (
    sales_df['Customer_ID'].str.strip().str.upper()
)

print("Text cleaning completed!")

print("\nProducts:")
print(sales_df['Product'].unique())

print("\nRegions:")
print(sales_df['Region'].unique())

print("\nCustomer IDs:")
print(sales_df['Customer_ID'].head())
```

```python
# ==========================================
# STEP 7: FEATURE ENGINEERING
# ==========================================

# Revenue per unit
sales_df['Revenue_Per_Unit'] = (
    sales_df['Total_Sales'] / sales_df['Quantity'].replace(0, np.nan)
)

# Categorize purchase quantity
sales_df['Quantity_Category'] = pd.cut(
    sales_df['Quantity'],
    bins=[-np.inf, 2, 5, np.inf],
    labels=['Low', 'Medium', 'High']
)

# Categorize sales value using median
median_sales = sales_df['Total_Sales'].median()

sales_df['Sales_Category'] = np.where(
    sales_df['Total_Sales'] >= median_sales,
    'High Value',
    'Low Value'
)

print("Feature engineering completed!")

display(
    sales_df[
        ['Product', 'Quantity', 'Price', 'Total_Sales',
         'Revenue_Per_Unit', 'Quantity_Category',
         'Sales_Category']
    ].head(10)
)
```

```python
# ==========================================
# STEP 8: MISSING VALUE ANALYSIS
# ==========================================

print("SALES DATASET")
print(sales_df.isnull().sum())

print("\nCUSTOMER DATASET")
print(customer_df.isnull().sum())

print("\nMissing value percentages:")
print((sales_df.isnull().mean() * 100).round(2))
```

```python
# ==========================================
# STEP 9: FINAL DATA VALIDATION
# ==========================================

print("=" * 60)
print("FINAL DATA QUALITY REPORT")
print("=" * 60)

print("\nSales Dataset Shape:", sales_df.shape)

print("\nData Types:")
print(sales_df.dtypes)

print("\nDuplicate Records:", sales_df.duplicated().sum())

print("\nInvalid Dates:", sales_df['Date'].isnull().sum())

print("\nInvalid Sales Calculations:",
      (sales_df['Total_Sales'] !=
       sales_df['Quantity'] * sales_df['Price']).sum())

print("\nUnique Products:", sales_df['Product'].nunique())

print("Unique Regions:", sales_df['Region'].nunique())

print("Unique Customers:", sales_df['Customer_ID'].nunique())

print("\nFinal validation completed!")
```

```python
# Save cleaned sales dataset

sales_df.to_csv(
    "cleaned_sales_data.csv",
    index=False
)

print("Cleaned dataset exported successfully!")
```

```python
# ==========================================
# DAY 3: CUSTOMER ANALYSIS
# STEP 1: CUSTOMER PURCHASE ANALYSIS
# ==========================================

customer_analysis = sales_df.groupby('Customer_ID').agg(
    Total_Spending=('Total_Sales', 'sum'),
    Total_Orders=('Total_Sales', 'count'),
    Total_Quantity=('Quantity', 'sum'),
    Average_Order_Value=('Total_Sales', 'mean')
).reset_index()

customer_analysis = customer_analysis.sort_values(
    by='Total_Spending',
    ascending=False
)

print("=" * 60)
print("CUSTOMER PURCHASE ANALYSIS")
print("=" * 60)

display(customer_analysis.head(10))

print("\nTotal Unique Customers:",
      customer_analysis['Customer_ID'].nunique())

print("Average Customer Spending:",
      customer_analysis['Total_Spending'].mean())
```

```python
# ==========================================
# TOP 10 CUSTOMERS
# ==========================================

top_customers = customer_analysis.head(10)

print("TOP 10 CUSTOMERS BY REVENUE")

display(top_customers)
```

```python
# ==========================================
# CUSTOMER LIFETIME VALUE
# ==========================================

customer_analysis['Customer_Lifetime_Value'] = (
    customer_analysis['Total_Spending']
)

print("CUSTOMER VALUE ANALYSIS")

display(
    customer_analysis[
        ['Customer_ID', 'Customer_Lifetime_Value']
    ].head(10)
)

print("\nAverage Customer Lifetime Revenue:",
      customer_analysis['Customer_Lifetime_Value'].mean())

print("Highest Customer Lifetime Revenue:",
      customer_analysis['Customer_Lifetime_Value'].max())
```

```python
# ==========================================
# PURCHASE FREQUENCY ANALYSIS
# ==========================================

print("CUSTOMER PURCHASE FREQUENCY")

print("\nAverage Orders Per Customer:",
      customer_analysis['Total_Orders'].mean())

print("\nMost Frequent Customers:")

display(
    customer_analysis.sort_values(
        by='Total_Orders',
        ascending=False
    ).head(10)
)
```

```python
# ==========================================
# STEP 5: REGIONAL SALES ANALYSIS
# ==========================================

regional_analysis = sales_df.groupby('Region').agg(
    Total_Revenue=('Total_Sales', 'sum'),
    Total_Orders=('Total_Sales', 'count'),
    Total_Quantity=('Quantity', 'sum'),
    Unique_Customers=('Customer_ID', 'nunique')
).reset_index()

regional_analysis = regional_analysis.sort_values(
    by='Total_Revenue',
    ascending=False
)

print("=" * 60)
print("REGIONAL SALES PERFORMANCE")
print("=" * 60)

display(regional_analysis)

print("\nHighest Revenue Region:")
print(regional_analysis.iloc[0]['Region'])

print("\nHighest Revenue:")
print(regional_analysis.iloc[0]['Total_Revenue'])
```

```python
# ==========================================
# STEP 6: CUSTOMER SEGMENTATION
# ==========================================

q1 = customer_analysis['Total_Spending'].quantile(0.33)
q2 = customer_analysis['Total_Spending'].quantile(0.66)

def categorize_customer(spending):

    if spending >= q2:
        return 'High Value'

    elif spending >= q1:
        return 'Medium Value'

    else:
        return 'Low Value'

customer_analysis['Customer_Segment'] = (
    customer_analysis['Total_Spending'].apply(categorize_customer)
)

print("CUSTOMER SEGMENTATION COMPLETED")

display(
    customer_analysis[
        ['Customer_ID', 'Total_Spending', 'Customer_Segment']
    ].head(15)
)

print("\nCustomer Segment Distribution:")

print(customer_analysis['Customer_Segment'].value_counts())
```

```python
# ==========================================
# STEP 7: CUSTOMER SEGMENT PERFORMANCE
# ==========================================

segment_analysis = customer_analysis.groupby(
    'Customer_Segment'
).agg(
    Total_Customers=('Customer_ID', 'count'),
    Total_Revenue=('Total_Spending', 'sum'),
    Average_Spending=('Total_Spending', 'mean')
).reset_index()

segment_analysis['Revenue_Contribution_Percent'] = (
    segment_analysis['Total_Revenue'] /
    segment_analysis['Total_Revenue'].sum()
) * 100

display(segment_analysis)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=customer_analysis.head(10),
    x='Customer_ID',
    y='Total_Spending'
)

plt.title('Top 10 Customers by Revenue')
plt.xlabel('Customer ID')
plt.ylabel('Total Spending')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

segment_counts = customer_analysis[
    'Customer_Segment'
].value_counts()

plt.pie(
    segment_counts,
    labels=segment_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Customer Segmentation Distribution')
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# DAY 4: SALES PATTERN ANALYSIS
# STEP 1: MONTHLY SALES AGGREGATION
# ==========================================

monthly_sales = sales_df.groupby(
    sales_df['Date'].dt.to_period('M')
).agg(
    Total_Revenue=('Total_Sales', 'sum'),
    Total_Quantity=('Quantity', 'sum'),
    Total_Transactions=('Total_Sales', 'count'),
    Average_Order_Value=('Total_Sales', 'mean')
).reset_index()

monthly_sales['Date'] = monthly_sales['Date'].astype(str)

print("=" * 60)
print("MONTHLY SALES PERFORMANCE")
print("=" * 60)

display(monthly_sales)
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_sales,
    x='Date',
    y='Total_Revenue',
    marker='o',
    linewidth=2
)

plt.title('Monthly Revenue Trend', fontsize=16)
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 3: BEST-SELLING PRODUCTS
# ==========================================

product_analysis = sales_df.groupby('Product').agg(
    Total_Quantity_Sold=('Quantity', 'sum'),
    Total_Revenue=('Total_Sales', 'sum'),
    Total_Orders=('Total_Sales', 'count'),
    Average_Price=('Price', 'mean')
).reset_index()

product_analysis = product_analysis.sort_values(
    by='Total_Quantity_Sold',
    ascending=False
)

print("PRODUCT PERFORMANCE ANALYSIS")

display(product_analysis)

print("\nBest-Selling Product:")

print(product_analysis.iloc[0]['Product'])
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=product_analysis.sort_values(
        by='Total_Revenue',
        ascending=False
    ),
    x='Product',
    y='Total_Revenue'
)

plt.title('Product-wise Revenue Performance')
plt.xlabel('Product')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 5: REGIONAL SALES TRENDS
# ==========================================

regional_monthly_sales = sales_df.pivot_table(
    index='Month_Name',
    columns='Region',
    values='Total_Sales',
    aggfunc='sum',
    fill_value=0
)

month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

regional_monthly_sales = regional_monthly_sales.reindex(
    [m for m in month_order if m in regional_monthly_sales.index]
)

display(regional_monthly_sales)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=regional_analysis,
    x='Region',
    y='Total_Revenue'
)

plt.title('Regional Revenue Comparison')
plt.xlabel('Region')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 7: MONTH-OVER-MONTH GROWTH
# ==========================================

monthly_sales['Revenue_Growth_Percent'] = (
    monthly_sales['Total_Revenue'].pct_change() * 100
)

display(monthly_sales)

print("\nHighest Revenue Month:")
print(
    monthly_sales.loc[
        monthly_sales['Total_Revenue'].idxmax(), 'Date'
    ]
)

print("\nLowest Revenue Month:")
print(
    monthly_sales.loc[
        monthly_sales['Total_Revenue'].idxmin(), 'Date'
    ]
)
```

```python
# ==========================================
# SALES PIVOT TABLE
# ==========================================

sales_pivot = pd.pivot_table(
    sales_df,
    values='Total_Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0,
    margins=True
)

print("REGION-WISE PRODUCT SALES PIVOT TABLE")

display(sales_pivot)
```

```python
# AND condition
high_value_sales = sales_df[
    (sales_df['Total_Sales'] > sales_df['Total_Sales'].median()) &
    (sales_df['Quantity'] >= 3)
]

print("HIGH VALUE SALES WITH QUANTITY >= 3")

display(high_value_sales.head(10))

# OR condition
selected_regions = sales_df[
    (sales_df['Region'] == 'North') |
    (sales_df['Region'] == 'South')
]

print("\nSALES FROM NORTH OR SOUTH")

display(selected_regions.head(10))
```

```python
# ==========================================
# STEP 5: CUSTOMER RETENTION ANALYSIS
# ==========================================

customer_frequency = sales_df.groupby('Customer_ID').agg(
    Total_Orders=('Total_Sales', 'count'),
    Total_Revenue=('Total_Sales', 'sum')
).reset_index()

# Identify repeat customers
customer_frequency['Customer_Type'] = np.where(
    customer_frequency['Total_Orders'] > 1,
    'Repeat Customer',
    'One-Time Customer'
)

# Calculate retention metrics
total_customers = len(customer_frequency)

repeat_customers = (
    customer_frequency['Total_Orders'] > 1
).sum()

retention_rate = (
    repeat_customers / total_customers
) * 100

print("=" * 60)
print("CUSTOMER RETENTION ANALYSIS")
print("=" * 60)

print("Total Customers:", total_customers)
print("Repeat Customers:", repeat_customers)
print("One-Time Customers:", total_customers - repeat_customers)

print(f"Repeat Customer Rate: {retention_rate:.2f}%")

display(customer_frequency.head(10))
```

```python
# ==========================================
# CUSTOMER PRODUCT AFFINITY
# ==========================================

customer_product = sales_df.pivot_table(
    index='Customer_ID',
    columns='Product',
    values='Quantity',
    aggfunc='sum',
    fill_value=0
)

# Convert quantities into purchase indicators
purchase_matrix = (customer_product > 0).astype(int)

# Calculate product co-occurrence
product_affinity = purchase_matrix.T.dot(
    purchase_matrix
)

print("CUSTOMER PRODUCT AFFINITY MATRIX")

display(product_affinity)
```
